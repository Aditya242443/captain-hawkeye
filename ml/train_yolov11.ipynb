```python
import os

if not os.path.exists('/content/train/images'):
    !unzip -q "/content/drive/MyDrive/SIH Dataset/Plate Number.v7i.yolov11.zip" -d /content/
    print("Extracted.")
else:
    print("Dataset already present, skipping extraction.")

print("train/images exists:", os.path.exists('/content/train/images'))
print("valid/images exists:", os.path.exists('/content/valid/images'))
print("test/images exists:", os.path.exists('/content/test/images'))
```

```python
yaml_content = """train: /content/train/images
val: /content/valid/images
test: /content/test/images

nc: 1
names: ['license_plate']
"""

with open('/content/data.yaml', 'w') as f:
    f.write(yaml_content)

print("data.yaml written:")
!cat /content/data.yaml
```

```python
!pip install -q ultralytics

import torch
print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("No GPU detected")
```

```python
from ultralytics import YOLO

model = YOLO('yolo11n.pt')

results = model.train(
    data='/content/data.yaml',
    epochs=20,
    imgsz=640,
    batch=16,
    project='/content/drive/MyDrive/SIH Dataset/runs',
    name='plate_detector_final',
    save_period=1,   # checkpoint after every epoch
    patience=20
)
```

```python
from ultralytics import YOLO
model = YOLO('/content/drive/MyDrive/SIH Dataset/runs/plate_detector_final/weights/best.pt')

results = model.predict('/content/valid/images', save=True, conf=0.5)
```
